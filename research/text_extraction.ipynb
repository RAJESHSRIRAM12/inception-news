```python
import feedparser

feed = feedparser.parse("https://rss.dw.com/atom/rss-en-all")
```

```python
import trafilatura as tr
#### Entries is where everything is stored ######
number_of_entries = len(feed.entries)
print(number_of_entries)

for entries in feed.entries:
    link = entries.link
    downloaded = tr.fetch_url(link)
    result = tr.extract(downloaded)
```
